```python
import pandas as pd

df = pd.read_csv("results/enhanced_news.csv")

df["real_topic"].value_counts()
```

```python
df = df[df["real_topic"] != "general"]

from sklearn.metrics import accuracy_score

acc = accuracy_score(df["real_topic"], df["topics"])
print("Accuracy:", acc)
```

```python
from sklearn.metrics import classification_report

print(classification_report(df["real_topic"], df["topics"]))
```

```python
df["correct"] = df["real_topic"] == df["topics"]

print(df["correct"].value_counts(normalize=True))
```
